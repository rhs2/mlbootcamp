```python
#line plot from the csv data
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('data/data2.csv')
ages = df['Age']
dev_salaries = df["All_Devs"]
py_salaries= df['Python']
js_salaries= df['JavaScript']
plt.plot(ages, py_salaries, color= '#444444', linestyle ='--' , label= 'Python' )
plt.plot(ages, dev_salaries, label = 'All dev' )
plt.legend()
plt.title("Median salary usd by age ")
plt.xlabel('Ages')
plt.ylabel('Median Salary')
plt.tight_layout()
plt.show()
```

```python
#fill on the line plot using alpha
#line plot from the csv data
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('data/data2.csv')
ages = df['Age']
dev_salaries = df["All_Devs"]
py_salaries= df['Python']
js_salaries= df['JavaScript']
plt.plot(ages, dev_salaries, color= '#444444', linestyle ='--' , label= 'All devs' )
plt.plot(ages, py_salaries, label = 'Python' )
plt.fill_between(ages, py_salaries, alpha = 0.25 )
plt.legend()
plt.title("Median salary usd by age ")
plt.xlabel('Ages')
plt.ylabel('Median Salary')
plt.tight_layout()
plt.show()
```

```python
#only fills with conditions with different colors
#fill between overall median 
#fill on the line plot using alpha
#line plot from the csv data
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('data/data2.csv')
ages = df['Age']
dev_salaries = df["All_Devs"]
py_salaries= df['Python']
js_salaries= df['JavaScript']
plt.plot(ages, dev_salaries, color= '#444444', linestyle ='--' , label= 'All devs' )
plt.plot(ages, py_salaries, label = 'Python' )
overall_median = 57287
plt.fill_between(ages, py_salaries, overall_median,where=(py_salaries > overall_median),
                interpolate=True, alpha = 0.25, )
plt.legend()
plt.title("Median salary usd by age ")
plt.xlabel('Ages')
plt.ylabel('Median Salary')
plt.tight_layout()
plt.show()
```
