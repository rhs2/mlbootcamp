```python
#create a basic pie chart it will be like a pie and slices of the pie
#adding labels to pie chart
#adding wedgeprops
from matplotlib import pyplot as plt
plt.style.use('fivethirtyeight')
slices =[60,40]
labels = ['sixty', 'forty']
plt.pie(slices, labels= labels, wedgeprops={'edgecolor': 'black'})

plt.title("Pie chart fot the data")
plt.tight_layout()
plt.show()

#hexcolor---->color
#blue ---> #008fd5
#red ---> #fc4f30
#yellow ---> #e5ae37
#green ---> #6d904f
```

```python
#create a basic pie chart it will be like a pie and slices of the pie
#adding labels to pie chart
#adding wedgeprops
#specify color 
from matplotlib import pyplot as plt
plt.style.use('fivethirtyeight')
slices =[60,40, 30,20]
labels = ['sixty', 'forty', 'extra1', 'extra2']
colors = ['#008fd5','#fc4f30', '#e5ae37','#6d904f' ]
plt.pie(slices, labels= labels, colors = colors, wedgeprops={'edgecolor': 'black'})

plt.title("Pie chart fot the data")
plt.tight_layout()
plt.show()

#hexcolor---->color
#blue ---> #008fd5
#red ---> #fc4f30
#yellow ---> #e5ae37
#green ---> #6d904f
```

```python
#real world example with 5 slices
#explode argument to offset the slices #explode python
#Language Popularity
#adding a shadow to the chart
explode = [0,0,0,.1,0]
slices = [59219, 55466, 47544, 36443, 35917]
labels = ['JavaScript', 'HTML/CSS', 'SQL', 'Python', 'Java']
colors = ['#008fd5','#fc4f30', '#e5ae37','#6d904f','#444444' ]
plt.pie(slices, labels=labels, colors = colors, 
        explode = explode,shadow= True,
         wedgeprops={'edgecolor': 'black'})

plt.title("Pie chart fot the data")
plt.tight_layout()
plt.show()
#it is not recommended to create with a lots of data, use pie chart when you have 5/6 data 
#if there is more then use bar plot
#.1 means it is 10% exploded from the radius
```

```python
#rotate the pie
#real world example with 5 slices
#explode argument to offset the slices #explode python
#Language Popularity
#adding a shadow to the chart
explode = [0,0,0,.1,0]
slices = [59219, 55466, 47544, 36443, 35917]
labels = ['JavaScript', 'HTML/CSS', 'SQL', 'Python', 'Java']
colors = ['#008fd5','#fc4f30', '#e5ae37','#6d904f','#444444' ]
plt.pie(slices, labels=labels, colors = colors, 
        explode = explode,shadow= True, startangle=90,
         wedgeprops={'edgecolor': 'black'})

plt.title("Pie chart fot the data")
plt.tight_layout()
plt.show()
#it is not recommended to create with a lots of data, use pie chart when you have 5/6 data 
#if there is more then use bar plot
#.1 means it is 10% exploded from the radius
#start the js at 90 degree

```

```python
#display the percentage 
#rotate the pie
#real world example with 5 slices
#explode argument to offset the slices #explode python
#Language Popularity
#adding a shadow to the chart
explode = [0,0,0,.1,0]
slices = [59219, 55466, 47544, 36443, 35917]
labels = ['JavaScript', 'HTML/CSS', 'SQL', 'Python', 'Java']
colors = ['#008fd5','#fc4f30', '#e5ae37','#6d904f','#444444' ]
plt.pie(slices, labels=labels, colors = colors, 
        explode = explode,shadow= True, startangle=90,
        autopct='%1.1f%%',
        wedgeprops={'edgecolor': 'black'})

plt.title("Pie chart fot the data")
plt.tight_layout()
plt.show()
#it is not recommended to create with a lots of data, use pie chart when you have 5/6 data 
#if there is more then use bar plot
#.1 means it is 10% exploded from the radius
#start the js at 90 degree
#creating auto percentage ----> autopct='%1.1f%%'

```
