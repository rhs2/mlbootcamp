```python
import pandas as pd
import matplotlib.pyplot as plt
plt.style.use("seaborn-v0_8")
df = pd.read_csv('data/data_subplots.csv')
ages = df['Age']
all_devs_sal= df['All_Devs']
py_dev_sal = df['Python']
js_dev_sal= df["JavaScript"]

plt.plot(ages, all_devs_sal, color= 'red', linestyle = "--", label= 'all devs')
plt.plot(ages, py_dev_sal, color= 'black',linestyle = "--",label= 'all devs')
plt.plot(ages, js_dev_sal, color= 'blue', linestyle = "--",label= 'all devs')
plt.title('Median salary usd by aage')
plt.xlabel('ages')
plt.ylabel('Median sal usd')
plt.tight_layout()
plt.legend()
plt.show()
```

```python
#sharing x 
# #broken up the plots into different plots

# #initiate figure with ax 

import pandas as pd
import matplotlib.pyplot as plt
plt.style.use("seaborn-v0_8")
df = pd.read_csv('data/data_subplots.csv')
ages = df['Age']
all_devs_sal= df['All_Devs']
py_dev_sal = df['Python']
js_dev_sal= df["JavaScript"]

fig,(ax1,ax2) = plt.subplots(nrows=2, ncols=1, sharex= True)

ax1.plot(ages, all_devs_sal, color= 'red', linestyle = "--", label= 'all devs')
ax2.plot(ages, py_dev_sal, color= 'black',linestyle = "--",label= 'python ')
ax2.plot(ages, js_dev_sal, color= 'blue', linestyle = "--",label= 'js')

ax1.legend()
ax1.set_title('Median salary usd by age')
ax1.set_ylabel('Median sal usd')

ax2.legend()
ax2.set_xlabel('ages')
ax2.set_ylabel('Median sal usd')
plt.tight_layout()
plt.show()
```
