```python
#working with real world example to see trend on views and likes
import pandas as pd
from matplotlib import pyplot as plt
df = pd.read_csv('data/data_scatter.csv')
v_c= df['view_count']
likes = df['likes']
ratio= df['ratio']
plt.style.use('seaborn-v0_8')

plt.scatter(v_c, likes, edgecolors='black',linewidth=1, alpha= 0.75)
plt.title('trending youtube videos ')
plt.xlabel('view counts of videos')
plt.ylabel('likes of the videos')
plt.tight_layout()
plt.show()
```

```python
#log scale with scatter plot
#working with real world example to see trend on views and likes
import pandas as pd
from matplotlib import pyplot as plt
df = pd.read_csv('data/data_scatter.csv')
v_c= df['view_count']
likes = df['likes']
ratio= df['ratio']
plt.style.use('seaborn-v0_8')

plt.scatter(v_c, likes, edgecolors='black',linewidth=1, alpha= 0.75)
plt.xscale('log') #use the log scale to see the correlation better
plt.yscale('log')
plt.title('trending youtube videos ')
plt.xlabel('view counts of videos')
plt.ylabel('likes of the videos')
plt.tight_layout()
plt.show()
```

```python
#adding colors
#log scale with scatter plot
#working with real world example to see trend on views and likes
import pandas as pd
from matplotlib import pyplot as plt
df = pd.read_csv('data/data_scatter.csv')
v_c= df['view_count']
likes = df['likes']
ratio= df['ratio']
plt.style.use('seaborn-v0_8')

plt.scatter(v_c, likes, c= ratio, cmap= 'summer',edgecolors='black',linewidth=1, alpha= 0.75)
plt.xscale('log') #use the log scale to see the correlation better
plt.yscale('log')
cbar = plt.colorbar()
cbar.set_label('Ratio')
plt.title('trending youtube videos ')
plt.xlabel('view counts of videos')
plt.ylabel('likes of the videos')
plt.tight_layout()
plt.show()
```
