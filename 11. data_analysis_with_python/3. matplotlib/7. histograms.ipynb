```python
#basic histograms
import pandas as pd
import matplotlib.pyplot as plt
plt.style.use('fivethirtyeight')
ages = [18,19,21,25,26,26,30,32,38,45,55]
plt.hist(ages)
plt.title("Ages of respondent")
plt.xlabel('ages')
plt.ylabel('Total respondent')
plt.tight_layout()
plt.show()
```

```python
# adding bins explicitly pass in int value and hist will make it by automatically
#basic histograms
#5 different bins
import pandas as pd
import matplotlib.pyplot as plt
plt.style.use('fivethirtyeight')
bins = 5
ages = [18,19,21,25,26,26,30,32,38,45,55]
plt.hist(ages, bins= bins)
plt.title("Ages of respondent")
plt.xlabel('ages')
plt.ylabel('Total respondent')
plt.tight_layout()
plt.show()
```

```python
#we can add bins explicitly by telling the number range passing through a list
#also adding edgecolor to make the bins border clear
import pandas as pd
import matplotlib.pyplot as plt
plt.style.use('fivethirtyeight')
bins = [10,20,30,50,60]
ages = [18,19,21,25,26,26,30,32,38,45,55]
plt.hist(ages, bins= bins, edgecolor= 'black')
plt.title("Ages of respondent")
plt.xlabel('ages')
plt.ylabel('Total respondent')
plt.tight_layout()
plt.show()
```

```python
##plot the hist into logarithmic
#looking at real words data
import pandas as pd
import matplotlib.pyplot as plt
plt.style.use('fivethirtyeight')
df = pd.read_csv('data/data3.csv')
# print(df.head(5))
ids = df['Responder_id']
ages= df['Age']
bins = [20,30,40,50,60,70,80,90,100]
plt.hist(ages, bins = bins, edgecolor= 'black', log= True)
plt.title("Ages of respondent")
plt.xlabel('ages')
plt.ylabel('Total respondent')
plt.tight_layout()
plt.show()
```

```python
#adding #axvline---> axis vertical line method
#add some aditional info , plot a vertical line of the median age
##plot the hist into logarithmic
#looking at real words data
import pandas as pd
import matplotlib.pyplot as plt
plt.style.use('fivethirtyeight')
df = pd.read_csv('data/data3.csv')
# print(df.head(5))
ids = df['Responder_id']
ages= df['Age']
bins = [20,30,40,50,60,70,80,90,100]
median_age= 29
color= '#fc4f30'
plt.hist(ages, bins = bins, edgecolor= 'black', log= True)
plt.axvline(median_age, color= color, label= 'Median age line' )
#axvline---> axis vertical line method
plt.legend()
plt.title("Ages of respondent")
plt.xlabel('ages')
plt.ylabel('Total respondent')
plt.tight_layout()
plt.show()
```
