```python
#plotting bar chart for languages and popularity
import csv
import numpy as np
from collections import Counter
from matplotlib import pyplot as plt

# Set style and figure size
plt.style.use('seaborn-v0_8')
plt.figure(figsize=(10, 8))

# Read and process data
with open('data/data.csv') as file:
    csv_reader = csv.DictReader(file)
    language_counter = Counter()
    for row in csv_reader:
        language_counter.update(row['LanguagesWorkedWith'].split(";"))

# Get the top 15 languages and reverse them for better display
language = []
popularity = []
for item in language_counter.most_common(15):
    language.append(item[0])
    popularity.append(item[1])
# Reverse lists so most popular is at top
language.reverse()
popularity.reverse()

# Create horizontal bar chart
bars = plt.barh(language, popularity, color='#444444', edgecolor='black')

# Customize the plot
plt.title("Most Popular Programming Languages", fontsize=14, pad=15)
plt.xlabel("Number of Developers", fontsize=12)
# No need to rotate labels in horizontal bar chart

# Add grid for better readability
plt.grid(axis='x', linestyle='--', alpha=0.7)

# Add value labels on end of each bar
for bar in bars:
    width = bar.get_width()
    plt.text(width, bar.get_y() + bar.get_height()/2,
             f'{int(width):,}',
             ha='left', va='center', fontsize=10)

# Adjust layout
plt.tight_layout()

plt.show()
```
