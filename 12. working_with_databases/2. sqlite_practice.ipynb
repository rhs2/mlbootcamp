2. Write a Python function to create a table named `employees` with columns `id` (integer), `name` (text), `age` (integer), and `department` (text) in the `test.db` database.

```python
def create_table():
    conn = sqlite3.connect('test.db')
    cursor= conn.cursor()
    cursor.execute('''
    CREATE TABLE IF NOT EXISTS employees(
                   id INTEGER PRIMARY KEY,
                   name TEXT NOT NULL,
                   age INTEGER,
                   department TEXT
                   )

    ''')

    conn.commit()
    conn.close()
    print("Table 'employees' created succesfully")
#test the functions
create_table()
```

### Assignment 2: Inserting Data

1. Write a Python function to insert a new employee into the `employees` table.

```python
def insert_employees(id, name, age, department):
    conn= sqlite3.connect('test.db')
    cursor = conn.cursor()
    cursor.execute('''
    INSERT INTO employees (id, name, age, department)
                   values (?,?,?,?)
    ''', (id, name, age, department))
    conn.commit()
    conn.close()
    print("Inserted data into employees succesfully")

#test the function
insert_employees(1,'rhs2', 30, 'Ml Engineer')
```

### Assignment 3: Querying Data

1. Write a Python function to fetch and display all records from the `employees` table.

```python
def fetch_all_employees():
    conn = sqlite3.connect('test.db')
    cursor = conn.cursor()
    cursor.execute('SELECT * FROM employees')
    records = cursor.fetchall()
    for r in records:
        print(r)
#test the function
fetch_all_employees()
```

2. Write a Python function to fetch and display all employees from a specific department.

```python
def fetch_all_employees_by_department(department):
    conn = sqlite3.connect('test.db')
    cursor = conn.cursor()
    cursor.execute('SELECT * FROM employees WHERE department = ?',(department,))
    records_d= cursor.fetchall()
    for r in records_d:
        print(r)

#test the function
fetch_all_employees_by_department('HR')
```

### Assignment 4: Updating Data

1. Write a Python function to update the department of an employee based on their `id`.

```python
def update_employees_department(employee_id, new_department):
    conn= sqlite3.connect('test.db')
    cursor = conn.cursor()
    cursor.execute("""
    UPDATE employees
                   SET department = ?
                   WHERE id =?
                   """, (new_department, employee_id))
    conn.commit()
    conn.close()
    print("Employee department updated succesfully")
#test the function
update_employees_department(1, 'Finance')
```

### Assignment 5: Deleting Data

1. Write a Python function to delete an employee from the `employees` table based on their `id`.

```python
def delete_employee(id):
    conn = sqlite3.connect('test.db')
    cursor = conn.cursor()
    cursor.execute('''
    DELETE from employees 
                   where id =?
    ''', (id,))
    conn.commit()
    conn.close()

#test the function
delete_employee(1)
#get the query of the employee
fetch_all_employees()
```
