```python
#different ways to format the plot, run th auto format and rotating the date
#using linestyle to connect the dots 
#basic plot date series
import pandas as pd
from datetime import datetime, timedelta
from matplotlib import dates as mpl_dates
import matplotlib.pyplot as plt
plt.style.use('seaborn-v0_8')

dates= [
    datetime(2019, 5, 24),
    datetime(2019, 5, 25),
    datetime(2019, 5, 26),
    datetime(2019, 5, 27),
    datetime(2019, 5, 28),
    datetime(2019, 5, 29),
    datetime(2019, 5, 30)
    
]
y = [0,1,3,4,6,5,7]
plt.plot_date(dates, y, linestyle = 'solid')
plt.gcf().autofmt_xdate()
plt.tight_layout()
plt.show() 
```

```python
#change the format of the date  formatting month-day-year
#date formatting links 
'''
https://docs.python.org/3/library/datetime.html#strftime-and-strptime-behavior
'''

#different ways to format the plot, run th auto format and rotating the date
#using linestyle to connect the dots 
#basic plot date series
import pandas as pd
from datetime import datetime, timedelta
from matplotlib import dates as mpl_dates
import matplotlib.pyplot as plt
plt.style.use('seaborn-v0_8')

dates= [
    datetime(2019, 5, 24),
    datetime(2019, 5, 25),
    datetime(2019, 5, 26),
    datetime(2019, 5, 27),
    datetime(2019, 5, 28),
    datetime(2019, 5, 29),
    datetime(2019, 5, 30)
    
]
y = [0,1,3,4,6,5,7]
plt.plot(dates, y, linestyle = 'solid')
plt.gcf().autofmt_xdate()
date_format = mpl_dates.DateFormatter('%b %d %y')
plt.gca().xaxis.set_major_formatter(date_format)
plt.tight_layout()
plt.show() 

```

```python
#working with large data csv 
#change the format of the date  formatting month-day-year
#date formatting links 
'''
https://docs.python.org/3/library/datetime.html#strftime-and-strptime-behavior
'''

#different ways to format the plot, run th auto format and rotating the date
#using linestyle to connect the dots 
#basic plot date series
import pandas as pd
from datetime import datetime, timedelta
from matplotlib import dates as mpl_dates
import matplotlib.pyplot as plt
df = pd.read_csv('data/data_datetime_series.csv')
price_date= df['Date']
price_close= df['Close']
plt.style.use('seaborn-v0_8')

plt.plot(price_date, price_close, marker= 'X', c= 'black')
plt.gcf().autofmt_xdate()
date_format = mpl_dates.DateFormatter('%b %d %y')
plt.gca().xaxis.set_major_formatter(date_format)
plt.tight_layout()
plt.show() 

```
